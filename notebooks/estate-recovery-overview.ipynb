# Medicaid estate recovery and the family home, 50 states and DC

This notebook loads `data/states.csv` and reproduces the headline counts in the README. Each row is one jurisdiction. The five method columns record whether Medicaid estate recovery can reach a house passing at the recipient's death by that method.

```python
import pandas as pd

df = pd.read_csv('../data/states.csv', dtype=str, keep_default_na=False)
METHODS = ['life_estate', 'lady_bird_deed', 'tod_deed', 'joint_tenancy', 'revocable_trust']
len(df)
```

## How far recovery reaches

`probate_only` states recover only from the probate estate. `expanded` states adopted the optional federal definition that reaches property passing outside probate.

```python
df['recovery_scope'].value_counts()
```

## Treatment of the house by transfer method

```python
pd.DataFrame({m: df[m].value_counts() for m in METHODS}).fillna(0).astype(int).T
```

## Probate-only is not the same as safe

Jurisdictions that limit recovery to the probate estate but can still reach a house in a revocable trust, because the trust code makes the trust liable when the probate estate is short.

```python
po = df[df['recovery_scope'] == 'probate_only']
po.loc[po['revocable_trust'] == 'reachable', ['code', 'name', 'revocable_trust_cite']]
```

## Transfer on death deed statutes by effective date

```python
tod = df[df['tod_deed_recognized'] == 'yes'][['code', 'name', 'tod_deed_effective', 'tod_deed_uniform_act', 'tod_deed']]
tod.sort_values('tod_deed_effective')
```
